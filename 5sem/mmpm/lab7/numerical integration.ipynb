### 1. Постановка задачи
$I = \int\limits_a^bf(x)dx$

$I = \underset{n \rightarrow \infty}{\lim} \sum\limits_{i = 1}^n f(\xi_i)(x_i - x_{i-1})$

$\xi_i \in [x_i - x_{i-1}]$

$[a, b] \rightarrow [x_0, x_1, \dots, x_n]$

#### Простые квадратурные формулы
##### Общая формула прямоугольников

$I \approx h\sum\limits_{i = 1}^{n}f(\xi_i)$

$x_i = x_0 + ih$

$h = \frac{b - a}{n}$

$\xi_i \in [x_i - x_{i-1}]$

##### Форула левых прямоугольников
$\xi_i = x_{i-1}$

$I \approx I_{пр}^{л} = h\sum\limits_{i = 1}^{n}f(x_{i - 1})$

##### Форула правых прямоугольников
$\xi_i = x_i$

$I \approx I_{пр}^{п} = h\sum\limits_{i = 1}^{n}f(x_i)$

###### Глобальная погрешность для формулы  левых и правых прямоугольников
$R_{пр} = \frac{b - a}{2}hf'(\xi) \rightarrow O(h)$

##### Форула средних прямоугольников
$\xi_i = x_i - \frac{h}{2}$

$I \approx I_{пр}^{ср} = h\sum\limits_{i = 1}^{n}f(x_i + \frac{h}{2})$

###### Глобальная погрешность для формулы  средних прямоугольников
$R_{пр} = \frac{b - a}{24}h^2f''(\xi) \rightarrow O(h^2)$


#### Квадратурные формула Ньютона-Котеса
$f(x) \approx L_n(x) = \sum\limits_{i = 0}^{n}f(x_i)\prod\limits_{j \neq i j = 0}^{n}\frac{x - x_j}{x_i - x_j}$

$\int\limits_a^b f(x)dx \approx \int\limits_a^bL(x) dx$

##### Квадратурная формула трапеций
$f(x) \approx L_1(x)$

$I \approx \frac{f(x_{i - 1}) + f(x_i)}{2}h$

###### Глобальная погрешность для формулы  трапеций
$R_{тр} = -\frac{b - a}{12}h^2f''(\xi), \xi \in [a,b] \rightarrow O(h^2)$


##### Квадратурная формула Симпсона
$f(x) \approx L_2(x)$

$I \approx \frac{h}{3}(f_{2i - 2} + 4f_{2i - 1} + f_{2_i})$

###### Глобальная погрешность для формулы  Симпсона
$R_{симпс} = -\frac{b - a}{180}h^2f^{IV}(\xi), \xi \in [a,b] \rightarrow O(h^4)$

```python
import numpy as np
from ipywidgets import interact, FloatRangeSlider, BoundedIntText
from matplotlib import pyplot as plt
```

Левые прямоугольники

```python
def riemann_sum_left(f, a, b, n):
    x = np.linspace(a, b, n + 1)
    y = f(x)
    h = (b - a) / n
    return np.sum(y[:-1] * h)
```

Правые прямоугольники

```python
def riemann_sum_right(f, a, b, n):
    x = np.linspace(a, b, n + 1)
    y = f(x)
    h = (b - a) / n
    return np.sum(y[1:] * h)
```

Средние прямоугольники

```python
def riemann_sum_midpoint(f, a, b, n):
    x = np.linspace(a, b, n + 1)
    y = f((x[1:] + x[:-1]) / 2)
    h = (b - a) / n
    return np.sum(y * h)
```

Трапеции

```python
def riemann_sum_trapezoid(f, a, b, n):
    x = np.linspace(a, b, n + 1)
    y = f(x)
    h = (b - a) / n
    return np.sum((y[1:] + y[:-1]) / 2 * h)
```

Симпсон

```python
def simpson(f, a, b, n):
    x = np.linspace(a, b, n + 1)
    y = f(x)
    h = (b - a) / n

    return h / 3 * (y[0] + y[-1] + 4 * np.sum(y[1:-1:2]) + 2 * np.sum(y[2:-2:2]))
```

```python
@interact(
    a_b=FloatRangeSlider(
        value=[-0, np.pi],
        min=-10,
        max=10,
        step=0.1,
        description='$[a, b]$:',
    ),
    n=BoundedIntText(
        min=2,
        step=1,
        description='$n$',
    ),
    e_pow = BoundedIntText(
        min=1,
        step=1,
        description='$1e-$',
    ),
)
def show_plot(a_b, n, e_pow):
    f = lambda x: np.sin(x)
    
    a, b = a_b
    h = (b - a) / n
    eps = 10 ** (-e_pow)
    
    F = lambda x: -np.cos(x)
    I = lambda a, b: F(b) - F(a)

    f_d = lambda x: np.cos(x)
    f_2d = lambda x: -np.sin(x)
    f_4d = lambda x: np.sin(x)

    max_f_d = np.max(np.abs(f_d(np.linspace(a, b, 1000))))
    max_f_2d = np.max(np.abs(f_2d(np.linspace(a, b, 1000))))
    max_f_4d = np.max(np.abs(f_4d(np.linspace(a, b, 1000))))

    theoretical_error = {
        riemann_sum_left: h * (b - a) * max_f_d / 2,
        riemann_sum_right: h * (b - a) * max_f_d / 2,
        riemann_sum_midpoint: h ** 2 * (b - a) * max_f_2d / 24,
        riemann_sum_trapezoid: -1 * (b - a) * h ** 2 * max_f_2d / 12,
        simpson: -1 * (b - a) * h ** 2 * max_f_4d / 180
    }

    fig, ax = plt.subplots(5, 1, figsize=(5, 10))
    x_plot = np.linspace(a, b, 1000)
    
    for i in range(5):
        ax[i].plot(x_plot, f(x_plot), label="sin(x)")

    x = np.linspace(a, b, n + 1)
    y = f(x)
    h = (b - a) / n

    ax[0].bar(x[:-1] + h / 2, y[:-1], width=h, alpha=0.5)
    ax[0].set_title("Левые прямоугольники: {}".format(riemann_sum_left(f, a, b, n)))

    ax[1].bar(x[:-1] + h / 2, y[1:], width=h, alpha=0.5)
    ax[1].set_title("Правые прямоугольники: {}".format(riemann_sum_right(f, a, b, n)))

    ax[2].bar(x[:-1] + h / 2, f((x[1:] + x[:-1]) / 2), width=h, alpha=0.5)
    ax[2].set_title("Средние прямоугольники: {}".format(riemann_sum_midpoint(f, a, b, n)))

    ax[3].fill_between(x, y, alpha=0.5)
    ax[3].set_title("Трапеции: {}".format(riemann_sum_trapezoid(f, a, b, n)))

    x = np.linspace(a, b, n + 1)
    y = f(x)
    h = (b - a) / n
    for i in range(n):
        x_i = np.linspace(x[i], x[i + 1], 100)
        y_i = (y[i + 1] - y[i]) / h * (x_i - x[i]) + y[i]
        ax[4].fill_between(x_i, y_i, color="C0", alpha=0.5)
    ax[4].set_title("Симпсон: {}".format(simpson(f, a, b, n)))
    plt.tight_layout()
    
    data = []
    for method in [riemann_sum_left, riemann_sum_right, 
                   riemann_sum_midpoint, riemann_sum_trapezoid, 
                   simpson]:
        n = 2
        while True:
            n += 2
            err = abs(method(f, a, b, n) - method(f, a, b, n - 1)) 
            if err < eps:
                data.append([
                    method.__name__,
                    I(a, b),
                    method(f, a, b, n),
                    err,
                    theoretical_error[method],
                    n
                ])
                break
    
    fig2, ax2 = plt.subplots(figsize=(18, 9))


    ax2.table(
        cellText=data,
        colLabels=[
            r'Метод',
            r'$I_{точное}$',
            r'$I_{прибл}$',
            r'$|I_{точное} - I_{приближенное}|$',
            r'$|I_{теор}|$',
            r'$n$ узлов'
        ],
        loc="top"
    )
    ax2.axis("off")

    ax, ax2
```
