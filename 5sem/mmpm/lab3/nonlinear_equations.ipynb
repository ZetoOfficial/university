### 1. Постановка задачи
Дано уравнение $f(x) = 0$, где $f(x)$ определена в некотором интервале $[a, b]$

Корень уравнение - $\xi$, значение такое, что $f(\xi) = 0$

**Требуется найти приближенное значение $\xi$**

#### Метод бисекций
В середине отрезка $x_1=\frac{a+b}{2}$ определяется знак функции $f(x)$, затем выбирается та половина отрезка $[a, \frac{a+b}{2}]$ или $[\frac{a+b}{2},b]$, 
на концах которой функция $f(x)$ принимает значения разных знаков, и деление повторяется.

Итерации продолжаются до тех пор, пока длина половины отрезка $[a, b]$ больше заданной точности $\varepsilon$

Условие окончания:
($\frac{(b - a)}{2} \leq \varepsilon$)

#### Метод хорд
В этом методе нелинейная функция $f(x)$ на отделенном интервале 
$[a, b]$ заменяется линейной, в качестве которой берется хорда – прямая, стягивающая концы нелинейной функции. 

Эта хорда определяется как прямая, проходящая через точки с координатами $(a ,f(a))$ и $(b, f(b))$, т.е. делит отрезок $[a, b]$  пропорционально величинам ординат, $f(a)$ и $f(b)$. 

Имея уравнения хорды, можно найти её точку пересечения с осью Ox, которую принимают за новую границу интервала, где содержится корень. Выбирается тот интервал $(a, x_i)$ и $(x_i, b)$, на концах которого функция имеет разные знаки. Через эту точку с координатами $(x_i, f(x_i))$ и соответствующую границу предыдущего промежутка снова проводят хорду и повторяют ранее описанную операцию, пока не будет достигнута нужная точность: $|x_n - x_{n-1}| \leq \varepsilon$

**Метод применим только для монотонных функций**

Если $f(b)f''(b)$ > 0, то строящаяся на каждом этапе хорда имеет превый фиксированный конец и алгоритм выглядит следующим образом:
$x_{i+1} = x_i \frac{f(x_i)}{f(b) - f(x_i)}(b - x_i)$

При этом последовательность $x_1, x_2, \dots$   будет приближаться к корню слева, в качестве $x_0$ можно выбрать точку $a$.


Если $f(a)f''(a)$ > 0, то строящаяся на каждом этапе хорда имеет левый фиксированный конец и алгоритм выглядит следующим образом:
$x_{i+1} = x_i \frac{f(x_i)}{f(a) - f(x_i)}(a - x_i)$

При этом последовательность $x_1, x_2, \dots$   будет приближаться к корню справа, в качестве $x_0$ можно выбрать точку $b$.

Условие окончания итерационного процесса:
$|x_i - x_{i - 1}| \leq \varepsilon$

#### Метод Ньютона
Если есть некоторое приближение $x_i$ к решению $x^*$, то метод Ньютона 
аппроксимирует функцию $f(x)$ касательной к ее графику в точке $x_i$ 

В качестве начальной точки в зависимости от свойств функции берется или левая точка $х_0 = a$, если $f(a)f''(a) > 0$, 
или правая точка $x_0 = b$, если $f(b) f''(b) > 0$

Далее итерации вычисляем как Далее итерации вычисляем как
$x_{i+1} = x_i - \frac{f(x_i)}{f'(x_i)}$

Условие окончания итерационного процесса:
$|x_i - x_{i - 1}| \leq \varepsilon$

#### Метод секущих
Метод реализуется алгоритмом метода хорд, только $a$ и $b$ взяты с одной стороны от корня и не фиксируются. 
Для начала расчета надо задавать два начальных 
приближения $x_0$  и $x_1$.
Далее итерации вычислять как:
$x_{n+1} = x_n - \frac{(x_n - x_{n - 1})f(x_n)}{f(x_n) - f(x_{n-1})}$

Условие окончания итерационного процесса: $|x_{i + 1} - x_{i}| \leq \varepsilon$

#### Метод итераций
Дано уравнение $f(x) = 0$, заменим его равносильным уравнением $x = \varphi(x)$
Итерации образуются по правил
$x_{i+1} = \varphi(x_i), i = 0, 1, 2 \dots$

**Теорема (о сходимости метода итераций)**. 

Пусть функция $\phi(х)$ определена и дифференцируема на отрезке. $[a, b]$ 
Тогда, если существует правильная дробь $q$ такая, что 
$|\phi'(х)| \leq q \leq 1$ при $a<x<b$, то  
1) Процесс итераций $х_{i+1}=\phi(х_i), i=0,1,2, \dots$	сходится независимо от начального значения $х_0 \in [a, b]$
2) Предельное значение  
$\xi = \lim_{i \to \infty} x_i$
является единственным корнем уравнения $х=\phi(х)$ на отрезке $[a, b]$.


Условие окончания итерационного процесса: $|x_{i} - x_{i - 1}| \leq \frac{1 - q}{q}\varepsilon$

Не всегда получается легко и просто подобрать функцию $\varphi'(x)$. Тогда можно оценить значения производной следующим выражением:

$0 \leq \alpha \leq f'(x) \leq \gamma < \infty$ и воспользоваться универсальными формулами выбора функции $x=x-\lambda f(x)$, где:

a) $\lambda=\frac{1}{\gamma}$, обеспечивает монотонную сходимость со скоростью $q(\lambda) = 1-\frac{\alpha}{\gamma}$ 

b) $\lambda = \frac{2}{(\alpha+\gamma)}$, обеспечивает оптимальную скорость сходимости: $q(\lambda)=\frac{\gamma - \alpha}{\gamma + \alpha}$

### 2. Метод решения

```python
import numpy as np
import pandas as pd
from tabulate import tabulate
from IPython.display import display, HTML
from matplotlib import pyplot as plt

np.set_printoptions(precision=15, suppress=True)
  
def print_table(table, headers):
    display(HTML(
        tabulate(table, headers=headers, tablefmt="unsafehtml")
    ))

def centered_axis(ax):
    ax.spines['left'].set_position('zero')
    ax.spines['right'].set_color('none')
    ax.spines['bottom'].set_position('zero')
    ax.spines['top'].set_color('none')
    ax.xaxis.set_ticks_position('bottom')
    ax.yaxis.set_ticks_position('left')
```

Метод бисекций

```python
def bisection(f, dfdx, a, b, eps):
    x = (a + b) / 2

    hist = [x]
    while b - a > 2 * eps:
        x = (a + b) / 2
        a, b = (a, x) if f(a) * f(x) < 0 else (x, b)
        hist.append(x)

    return x, np.array(hist)
```

Метод хорд

```python
def chords(f, dfdx, a, b, eps):
    x_prev, p = (b, a) if f(a) * dfdx(a) > 0 else (a, b)

    x = x_prev - f(x_prev) / (f(p) - f(x_prev)) * (p - x_prev)
    
    hist = [x_prev, x]
    while abs(x - x_prev) > eps:
        x_prev, x = x, x - f(x) / (f(p) - f(x)) * (p - x)
        hist.append(x)

    return x, np.array(hist)
```

Метод Ньютона

```python
def newton(f, dfdx, a, b, eps):
    x_prev = a if f(a) * dfdx(a) > 0 else b
    x = x_prev - f(x_prev) / dfdx(x_prev)

    hist = [x_prev, x]
    while abs(x - x_prev) > eps:
        x, x_prev = x - f(x) / dfdx(x), x
        hist.append(x)

    return x, np.array(hist)
```

Метод секущих

```python
def secant(f, dfdx, a, b, eps):
    x_prev, x = (b, a) if f(a) * dfdx(a) > 0 else (a, b)
    x_next = x - f(x) / (f(x) - f(x_prev)) * (x - x_prev)

    hist = [x_prev, x, x_next]
    while abs(x_next - x) > eps:
        x_prev, x, x_next = x, x_next, x_next - f(x_next) / (f(x_next) - f(x)) * (x_next - x)
        hist.append(x_next)

    return x_next, np.array(hist)
```

Метод простых итераций

```python
def simple_iterations(phi, x0, eps):
    x_prev = a
    x = phi(x_prev)

    hist = [x_prev, x]
    while abs(x - x_prev) > eps:
        x_prev, x = x, phi(x)
        hist.append(x)

    return x, np.array(hist)
```

#### Задание 1

```python
f = lambda x: np.exp(-x) - x ** 3
dfdx = lambda x: -np.exp(-x) - 3 * x ** 2
x = np.linspace(-0.5, 1.5, 1000)

a = 0
b = 1.5

fig, ax = plt.subplots()
ax.set_xlim(a - 0.01, b + 0.01)
ax.set_ylim(f(a), f(b))
centered_axis(ax)
ax.plot(x, f(x))
ax.plot(x, dfdx(x))

plt.show()
```

На отрезке [0, 1.5] уравнение имеет единственный корень

**Метод бисекций**

```python
fig, ax = plt.subplots()
ax.set_xlim(a - 0.01, b + 0.01)
ax.set_ylim(f(a), f(b))
centered_axis(ax)
ax.plot(x, f(x))


xi, iters = bisection(f, dfdx, a, b, 1e-3)
for x_i in iters:
    ax.plot([x_i, x_i], [0, f(x_i)], '--', lw=0.5, color='gray')
    ax.plot([0, x_i], [f(x_i), f(x_i)], '--', lw=0.5, color='gray')
    ax.plot(x_i, f(x_i), 'ko', ms=2)
ax.plot(xi, f(xi), 'rx')

plt.show()
```

**Метод хорд**

```python
fig, ax = plt.subplots()
ax.set_xlim(a - 0.01, b + 0.01)
ax.set_ylim(f(a), f(b))
centered_axis(ax)
ax.plot(x, f(x))

xi, iters = chords(f, dfdx, a, b, 1e-3)
p = a if f(a) * dfdx(a) > 0 else b
ax.plot([p, p], [0, f(p)], '--', lw=0.5, color='gray')
ax.plot([0, p], [f(p), f(p)], '--', lw=0.5, color='gray')
for x_i in iters:
    ax.plot([p, x_i], [f(p), f(x_i)], '-', lw=0.5, color='gray')
    ax.plot([x_i, x_i], [0, f(x_i)], '--', lw=0.5, color='gray')
    ax.plot([0, x_i], [f(x_i), f(x_i)], '--', lw=0.5, color='gray')
    ax.plot(x_i, f(x_i), 'ko', ms=2)
ax.plot(xi, f(xi), 'rx')

plt.show()
```

**Метод Ньютона**

```python
fig, ax = plt.subplots()
ax.set_xlim(a - 0.01, b + 0.01)
ax.set_ylim(f(a), f(b))
centered_axis(ax)
ax.plot(x, f(x))


xi, iters = newton(f, dfdx, a, b, 1e-3)
for x_i, x_j in zip(iters, iters[1:]):
    ax.plot([x_i, x_i], [0, f(x_i)], '--', lw=0.5, color='gray')
    ax.plot([0, x_i], [f(x_i), f(x_i)], '--', lw=0.5, color='gray')
    ax.plot([x_i, x_j], [f(x_i), 0], '-', lw=0.5, color='gray')
    ax.plot(x_i, f(x_i), 'ko', ms=2)
ax.plot(xi, f(xi), 'rx')

plt.show()
```

**Метод секущих**

```python
fig, ax = plt.subplots()
ax.set_xlim(a - 0.01, b + 0.01)
ax.set_ylim(f(a), f(b))
centered_axis(ax)
ax.plot(x, f(x))

xi, iters = secant(f, dfdx, a, b, 1e-3)
for x_i, x_j in zip(iters, iters[1:]):
    ax.plot([x_i, x_i], [0, f(x_i)], '--', lw=0.5, color='gray')
    ax.plot([0, x_i], [f(x_i), f(x_i)], '--', lw=0.5, color='gray')
    ax.plot([x_i, x_j], [f(x_i), 0], '-', lw=0.5, color='gray')
    ax.plot(x_i, f(x_i), 'ko', ms=2)
ax.plot(xi, f(xi), 'rx')


plt.show()
```

```python
results = []
methods = [
    ('бисекций', bisection),
    ('хорд', chords),
    ('Ньютона', newton),
    ('секущих', secant)
]


for name, method in methods:
    for eps in [1e-3, 1e-5, 1e-8]:
        xi, iters = method(f, dfdx, a, b, eps)
        results.append([name, iters[0], xi, len(iters), eps, f(xi)])

print_table(
    results,
    headers=[r'Метод', 
             r'Начальное приближение $x_0$', 
             r'Полученный корень $\xi$', 
             r'Число итераций $k$',
             r'Величина погрешности $\varepsilon$',
             r'Проверка $f(\varepsilon)$']
)
```
