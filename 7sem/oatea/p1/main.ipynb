```python
import re
import random
```

2. Загрузка текста из txt файла

```python
with open('text.txt', 'r') as file:
    text = file.read()

print(f"{text=}")
```

3. Обработка текста:
- Предварительная обработка текста
- Убираем прямую речь и лишние символы, приводим текст к нижнему регистру

Долой кавычки

```python
text = re.sub(r'[«»"“”]', '', text)
text = text.lower()
print(f"{text=}")
```

Разбивка на предложения по . ! ?

И пустых символов (пробелы)

```python
sentences = re.split(r'[.!?]', text)
sentences = [sentence.strip() for sentence in sentences if sentence.strip()]
print(f"{sentences=}")
```

```python
def clean_word(word):
    return re.sub(r'[^\w\s]', '', word)

word_lists = []
for sentence in sentences:
    words = [clean_word(word) for word in sentence.split() if clean_word(word)]
    word_lists.append(words)
word_lists
```

Показать правильность обработки (вывести несколько предложений случайным образом).

```python
for _ in range(3):
    random_sentence = ' '.join(random.choice(word_lists))
    print(random_sentence)
```

4.	Вычислить и вывести:
- Кол-во слов в тексте (общее число и уникальные).

```python
total_words = [word for sentence in word_lists for word in sentence]

total_word_count = len(total_words)
print(f"{total_word_count=}")
unique_word_count = len(set(total_words))
print(f"{unique_word_count=}")
```

- Кол-во предложений в тексте (sentence_count)

```python
sentence_count = len(word_lists)
print(f"{sentence_count=}")
```

- Самое длинное (longest_word) и самое короткое (shortest_word) слово (за исключением служебных частей речи).


Список исключаемых слов в доп. файле.

```python
with open('excluded_words.txt', 'r', encoding='utf-8') as f:
    excluded_words = set(f.read().split())
```

```python
filtered_words = [word for word in total_words if word not in excluded_words]
longest_word = max(filtered_words, key=len)
shortest_word = min(filtered_words, key=len)
print(f"{longest_word=}")
print(f"{shortest_word=}")
```

- Средняя и медианная длины слов (в знаках) и предложений (в словах)


Средняя длина слов (предложений) – отношение суммы длин всех слов (предложений) к количеству слов (предложений).

Медианная длина слов (предложений) – длина слов (предложений), находящаяся в середине упорядоченного по возрастанию списка длин каждого слова (предложения) в тексте. 

Например, список слов такой: "не", "об", "что", "этом", "буду", "думал", "когда-нибудь", "рассказывать". Медианной длиной слова будет 4.

Слова (в знаках)

```python
word_lengths = [len(word) for word in total_words]
avg_word_length = sum(word_lengths) / len(word_lengths)
print(f"{avg_word_length=}")

sorted_word_lengths = sorted(word_lengths)
median_word_length = sorted_word_lengths[len(sorted_word_lengths) // 2]
print(f"{median_word_length=}")
```

Предложения (в словах)

```python
sentence_lengths = [len(sentence) for sentence in word_lists]
avg_sentence_length = sum(sentence_lengths) / len(sentence_lengths)
print(f"{avg_sentence_length=}")

sorted_sentence_lengths = sorted(sentence_lengths)
median_sentence_length = sorted_sentence_lengths[len(sorted_sentence_lengths) // 2]
print(f"{median_sentence_length=}")
```
